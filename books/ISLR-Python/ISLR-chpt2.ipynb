# ISLR Chapter Two: Linear Regression

Simple linear regression is where there are two parameters/coefficients, $\beta_{0}$ and $\beta_{1}$, that look like this in the model: $Y \approx \beta_{0} + \beta_{1}X$. It is necessary to estimate the parameters, and there are a number of ways to do that, the most common is least squares. The least squares method adds up the square of the risiduals, Residual Sum of Squares (RSS), with each residual being $e_{i} = y_{i} - \hat{y}_{i}$. When evaluating we are seeking to find how close $\hat{\beta_{0}}$ and $\hat{\beta_{1}}$ are to the population $\beta_{0}$ and $\beta_{1}$, we do this by working out the standard errors (SE). The formulas use standard deviation (SD), but as we don't know that, SD is substituted with Residual Standard Error (RSE) which uses RSS.

The SE can then be used to work out both confidence intervals and t-statistic. A 95% confidence interval gives the interval in which, with 95% probability, the true parameters lie. This is shown as $\hat{\beta_{1}} \pm 2 \dot SE(\hat{\beta_{1}})$, it is the same for the intercept parameter, just change $\beta_{1}$ for $\beta_{0}$. SE can also be used to test the null hypothesis, where H0: there is no link between X and Y -> $\beta_{1}=0$, or H1: there is a link between X and Y -> $\beta_{1}\ne 0$. To decide this, we need to prove that $\beta_{1}$ is sufficiently far from zero, by using a t-statistic: $t=\frac{\hat{\beta}_{1}}{SE(\hat{\beta}_{1})}$. To finish, a p-value is calculated that determines how probable getting the t value arrived at if $\beta_{1} = 0$, if it is lower than 0.05, or 0.01, then we say it is highly unlikely, and so there is an assocation between X and Y and we can reject the null hypothesis.

Once the null hypothesis has been rejected, we measure how well the model fits the data, using RSE and $R^{2}$. The RSE measures the lack of fit, the larger the RSE the less well the model fits, and the greater the deviation. For example, if the RSE was 3.26, in the exmaple that means a prediction may be 3260 away from the actual value. The $R^{2}$ provides us with the amount of variance explained by the model, the closer to 1, the more variance is explained.

Multiple linear regression has multiple X values with their own individual coefficients: $Y \approx \beta_{0} + \beta_{1}X_{1} + \beta_{2}X_{2} + \beta_{3}X_{3} + ... + \beta_{j}X_{j}$, where $\beta_{j}$ is "the average effect on Y of a single unit increase in $X_{j}$, holding all other predictors fixed". In the output of fitting a model, t-scores and p-values are created for each individual coefficient. However by chance some of these can be below 0.05 (e.g. there are 100 coefficients), and so we need to look to the F-statistic. If it is close to 1 then the null hypothesis may be true (though for models trained on large numbers of observations the F-statistic may be close to 1 but still have coefficients that effecy Y). It is necessary to use statistical software with the F-statistic, the number of observations, and the number of coefficients to compute the p-value and decide whether to reject the null hypothesis.
